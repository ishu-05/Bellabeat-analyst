```python
import pandas as pd
import numpy as np
import glob
import os
```

```python
data_path="C:/Desktop/Bellabeat-analyst/data/raw"
all_files=glob.glob(os.path.join(data_path, "*.csv"))

print(f"Found {len(all_files)} files")
all_files
```

```python
for f in all_files:
    df_temp = pd.read_csv(f)
    print(f"\n{os.path.basename(f)}")
    print("Shape:", df_temp.shape)
    print("Nulls:\n", df_temp.isnull().sum()[df_temp.isnull().sum() > 0])
```

```python
for f in all_files:
    df_temp = pd.read_csv(f, nrows=1)
    print(os.path.basename(f), "→", list(df_temp.columns))
```

```python
daily_activity = pd.read_csv(os.path.join(data_path, 'dailyActivity_merged.csv'))
sleep_day = pd.read_csv(os.path.join(data_path, 'sleepDay_merged.csv'))
hourly_steps = pd.read_csv(os.path.join(data_path, 'hourlySteps_merged.csv'))
hourly_intensities = pd.read_csv(os.path.join(data_path, 'hourlyIntensities_merged.csv'))
heart_rate = pd.read_csv(os.path.join(data_path, 'heartrate_seconds_merged.csv'))
minute_sleep = pd.read_csv(os.path.join(data_path, 'minuteSleep_merged.csv'))
weight_log = pd.read_csv(os.path.join(data_path, 'weightLogInfo_merged.csv'))
```

```python
daily_activity['ActivityDate']= pd.to_datetime(daily_activity['ActivityDate'], format="%m/%d/%Y")
daily_activity['ActivityDate'].dtype
daily_activity['ActivityDate'].head()
```

```python
sleep_day['SleepDay']=pd.to_datetime(sleep_day['SleepDay'], format='%m/%d/%Y %I:%M:%S %p')
sleep_day['SleepDay']=sleep_day['SleepDay'].dt.date
sleep_day['SleepDay'].dtype
```

```python
hourly_steps['ActivityHour']=pd.to_datetime(hourly_steps['ActivityHour'],format='%m/%d/%Y %I:%M:%S %p')
hourly_steps['ActivityHour'].dtype
```

```python
hourly_intensities['ActivityHour']=pd.to_datetime(hourly_intensities['ActivityHour'], format='%m/%d/%Y %I:%M:%S %p')
hourly_intensities['ActivityHour'].dtype
```

```python
minute_sleep['date']=pd.to_datetime(minute_sleep['date'],format= '%m/%d/%Y %I:%M:%S %p')
minute_sleep['date'].dtype
```

```python
heart_rate['Time']=pd.to_datetime(heart_rate['Time'], format='%m/%d/%Y %I:%M:%S %p')
heart_rate['Time'].dtype
```

```python
weight_log['Date']=pd.to_datetime(weight_log['Date'], format= '%m/%d/%Y %I:%M:%S %p')
weight_log['Date'].dtype
```

```python
for name, df in {
    'daily_activity': daily_activity,
    'sleep_day': sleep_day,
    'hourly_steps': hourly_steps,
    'hourly_intensities': hourly_intensities,
    'heart_rate': heart_rate,
    'minute_sleep': minute_sleep,
    'weight_log': weight_log
}.items():
    dupes = df.duplicated().sum()
    print(f"{name}: {dupes} duplicate rows")
```

```python
sleep_day = sleep_day.drop_duplicates()
minute_sleep = minute_sleep.drop_duplicates()
```

```python
print("sleep_day duplicates remaining:", sleep_day.duplicated().sum())
print("sleep_day new shape:", sleep_day.shape)

print("minute_sleep duplicates remaining:", minute_sleep.duplicated().sum())
print("minute_sleep new shape:", minute_sleep.shape)
```

```python
print("heart_rate key dupes:", heart_rate.duplicated(subset=['Id', 'Time']).sum())
print("minute_sleep key dupes:", minute_sleep.duplicated(subset=['Id', 'date']).sum())
print("hourly_steps key dupes:", hourly_steps.duplicated(subset=['Id', 'ActivityHour']).sum())
print("hourly_intensities key dupes:", hourly_intensities.duplicated(subset=['Id', 'ActivityHour']).sum())
print("weight_log key dupes:", weight_log.duplicated(subset=['Id', 'Date']).sum())
```
